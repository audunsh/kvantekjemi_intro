```python
import bubblebox as bb
import numpy as np
```

# Attractive gas

```python
b = bb.showcase.attractive_gas(4**3, size = (10,10,10))
b.vel_ *= .05
b.dt = 0.001

b.evince(realism=False)
```

```python
b.run(100000)
```

# Ideal gas

```python
b = bb.showcase.ideal_gas(4**3, size = (30,30,30))
b.vel_ *= .05
b.dt = 0.001

b.evince(realism=False)
```

```python
b.run(100000)
```

# Hard sphere gas

```python
b = bb.showcase.hard_sphere_gas(5**3, size = (15,15,15))
b.vel_ *= .05
b.evince(realism=False)
```

```python

#b.pos[2] = 10*((b.pos[2]-5.0)/10.0)**2.0 - 5.0
b.run(100000)
```

# Lennard-Jones gas

```python
b = bb.mdbox(6**3, size = (-10,-10,-10), vel = 500.0)
b.masses = np.random.randint(1,3,6**3)*.1
b.vel_ *= .05
b.evince(realism=False)
```

```python
b.set_temperature(100)
b.run(1000000)
```
